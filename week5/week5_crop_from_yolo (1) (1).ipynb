```python
# Step 1: Install & Imports
# NOTE: YOLOv8 is used only for inference here

!pip install ultralytics opencv-python

import os
import cv2
import numpy as np
from ultralytics import YOLO
```

```python
# Step 2: Load YOLOv8 Model
# TODO: Load your trained YOLOv8 model (Week 4)

# Example:
# model = YOLO("best.pt")

from ultralytics import YOLO

MODEL_PATH = "best_model.pt"   # path to your trained YOLOv8 model
model = YOLO(MODEL_PATH)

print("YOLOv8 model loaded successfully")
```

```python
# Step 3: Define Input & Output Paths

# Step 3: Define Input & Output Paths (using existing structure)

INPUT_DIRS = [
    "train/images",
    "val/images",
    "test/images"
]

OUTPUT_DIR = "cropped_damage"
os.makedirs(OUTPUT_DIR, exist_ok=True)

print("Input directories:", INPUT_DIRS)
print("Output directory:", OUTPUT_DIR)

```

```python
# Step 4: Run Detection & Crop Bounding Boxes

crop_id = 0
CONF_THRESHOLD = 0.3   # adjust between 0.25–0.4 if needed

for INPUT_DIR in INPUT_DIRS:
    for img_name in os.listdir(INPUT_DIR):
        img_path = os.path.join(INPUT_DIR, img_name)

        # Read image
        image = cv2.imread(img_path)
        if image is None:
            continue

        # YOLO inference
        results = model(image, conf=CONF_THRESHOLD)

        for r in results:
            if r.boxes is None:
                continue

            for box in r.boxes:
                x1, y1, x2, y2 = map(int, box.xyxy[0])

                # Safety checks
                if x2 <= x1 or y2 <= y1:
                    continue

                crop = image[y1:y2, x1:x2]

                if crop.size == 0:
                    continue

                crop_id += 1
                save_path = os.path.join(OUTPUT_DIR, f"crop_{crop_id}.jpg")
                cv2.imwrite(save_path, crop)

print(f"✅ Total cropped damage images saved: {crop_id}")

```
